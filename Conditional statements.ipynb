```python
#ATM

balance = 50000
correct_pin = 1234

ATM="WELCOME TO AMT".center(120)
print(ATM.upper())

CARD="PLEASE INSERT YOUR CARD".center(120)
print(CARD.upper())

entered_pin = int(input("Enter your PIN: ".center(120)))

if entered_pin != correct_pin:
    print("Incorrect PIN. Transaction denied.".center(120))
else:
    amount = int(input("Enter amount to withdraw: ".center(120)))
    
    if amount % 100 != 0:
        print("Amount must be in multiples of 100.".center(120))
    elif amount > 20000:
        print("Cannot withdraw more than Rs.20000 per transaction.".center(120))
    elif amount > balance:
        print("Insufficient balance.".center(120))
    else:
        balance = balance - amount
        print(f"Transaction successful! Amount withdrawn: Rs.{amount}".center(120))
        print(f"Remaining balance: Rs.{balance}".center(120))
```

```python
# leaf year

year=int(input("enter the year"))

if year%4==0:
    if year%400==0 or year%100!=0:
        print(year,"is leap year")
    else:
        print(year,"is not leap year")
else:
    print(year,"is not leap year")
```
